```python
import re
def verificar_link(link):
  padrao_suspeito = r"(free|promo|ganhe|presente|cliqueaqui|bonus)"
  if re.search(padrao_suspeito, link.lower()):
    return "⚠ POSSIVELMENTE GOLPE. NÃO CLIQUE!"
  else:
    return "✔ Link parece seguro (mas sempre verifique!)"

link = input("Cole aqui o link que deseja verificar: ")
verificar_link(link)


```
